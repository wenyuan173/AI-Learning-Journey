## PCA主成分分析

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
n = 1000
pc1 = np.random.normal(0,1,n)
pc2 = np.random.normal(0,0.5,n)

f1 =pc1+pc2
f2 = pc1-pc2
f3= pc1+ np.random.normal(0,0.1,n)

x= np.vstack([f1,f2,f3]).T
print(x.shape)
```

```python
pca = PCA(n_components=2)
# 现在是保留2个特征值，也可以写成0-1的小数代表保留的信息量

pca_after = pca.fit_transform(x)
print(pca_after.shape)
```

```python
fig = plt.figure(figsize=(12,4))

ax1 = fig.add_subplot(121,projection='3d')
ax1.scatter(x[:,0],x[:,1],x[:,2],c="g")
ax1.set_title('Before PCA(3D)')
ax1.set_xlabel('F1')
ax1.set_ylabel('F2')
ax1.set_zlabel('F3')

ax2 = fig.add_subplot(122)
ax2.scatter(pca_after[:,0],pca_after[:,1],c="g")
ax2.set_title('After PCA(2D)')
ax2.set_xlabel('PC1')
ax2.set_ylabel('PC2')

plt.show()
```
