## 低方差过滤

```python
from sklearn.feature_selection import VarianceThreshold ##导入低方差过滤
import numpy as np
```

```python
a = np.random.randn(10000) #构造特征：生成10000个标准正态分布随机数
b = np.random.normal(loc=5, scale=0.1, size=10000)

print({np.mean(a)}, {np.var(a)})
print({np.mean(b)}, {np.var(b)})
```

```python
# 构造特征向量,先堆叠再转置
x= np.vstack((a, b)).T
print(x)
```

```python
# 低方差过滤
var_thresh = VarianceThreshold(0.1)
x = var_thresh.fit_transform(x)
print(x)
```
